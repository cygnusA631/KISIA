```python
options = webdriver.ChromeOptions()
options.add_argument("--no-sandbox") # 권한 이슈가 발생하지 않도록 함. (필수)
options.add_argument("--headless") # selenium의 gui를 사용하지 않음. 백그라운드 실행 (필수)
options.add_argument("--verbose") # 브라우저에서 실행하는 명령어에 대한 자세한 로그를 보여줌.
options.add_argument("--disable-gpu") # gpu 설정 해제
options.add_argument("--window-size=1920, 1200") # 크롤링하는 웹사이트의 크기를 1920*1200으로 고정 (반응형 대비)
options.add_argument("--disable-dev-shm-usage") # 메모리 이슈가 발생하지 않도록 함.
```

```python
driver = webdriver.Chrome(options=options)
print(driver)
```

```python
from time import sleep

from selenium.webdriver.common.by import By
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait
```

```python
url = "https://www.ediya.com/contents/drink.html#c"
```

```python
driver.get(url)
driver.implicitly_wait(0.5)
print(driver.page_source)
```

```python

coffee_checkbox = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.ID, "COFFEE"))
)
driver.execute_script("arguments[0].click();", coffee_checkbox)
sleep(2)
```

```python
while True:
    try:
        more_button = WebDriverWait(driver, 5).until(
            EC.element_to_be_clickable((By.CLASS_NAME, "line_btn"))
        )
        driver.execute_script("arguments[0].click();", more_button)
        print("더보기 버튼 클릭")
        sleep(2)
    except:
        print("더보기 버튼 더 이상 없음!")
        break

print("메뉴 모두 로드 완료!")
```

```python
#  결과 페이지 소스를 BeautifulSoup으로 파싱
html = driver.page_source
soup = BeautifulSoup(html, "html.parser")

#  메뉴 리스트 추출해서 출력
menu_items = soup.select("#menu_ul li")
print(f"총 {len(menu_items)}개의 메뉴가 로드됨!\n")

for item in menu_items:
    name = item.select_one(".menu_tt span")
    if name:
        print("-", name.get_text(strip=True))

driver.quit()
```
